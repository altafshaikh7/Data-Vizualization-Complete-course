Matplotlib Basic

subplot ek plotting concept hai jisme ek hi graphical canvas (figure) ko multiple chhote parts me divide kiya jata hai, aur har part me alag data visualize kiya jata hai.

Matlab:
ek figure (poora canvas)
uske andar multiple axes (individual plots)
har axes = ek subplot


subplot ka main goal:

1️⃣ Comparison

Ek hi screen pe multiple graphs compare karna
👉 jaise:

sales vs profit
temperature vs humidity

2️⃣ Space optimization

Alag-alag windows kholne ki jagah
→ ek hi figure me sab dikhao


3️⃣ Data storytelling

Ek hi visualization me multiple aspects dikhana
→ better understanding

Subplots in Matplotlin

subplots ek high-level function hai jo ek hi step me:

1. Figure create karta hai
2. Multiple Axes (subplots) create karta hai
3. Unhe structured grid me arrange karta hai

 Matlab:
figure + multiple plots = ek saath ready

 Purpose 
1️⃣ Clean code

Manual switching nahi karna padta (like subplot)

2️⃣ Better control

Har plot ko directly access kar sakte ho:
👉 ax[row, col]

3️⃣ Scalable

Large dashboards, multiple charts easy ho jaate hain

```python
plt.subplot(2, 2, 1)
plt.plot(x,y)
plt.subplot(2, 2, 4)
plt.plot(x,y)
plt.subplot(2, 2, 3)
plt.plot(y,x)
plt.subplot(2, 2, 2)
plt.plot(x**2,x**2)
```

Object Oriented Way

Object Oriented approach me tum directly “objects” (Figure aur Axes) ko control karte ho, instead of global plt functions.

Matlab:

> Tum bolte ho: “is specific plot pe ye draw karo”
> na ki: “jo current plot hai usme draw karo”

Why OO Approach BEST hai?
1️⃣ Full control

Har subplot pe alag settings laga sakte ho

2️⃣ No confusion

subplot() me kaunsa active hai → confusion hota hai
👉 OO me direct access

3️⃣ Scalable

Large projects me clean aur maintainable code


```python
fig = plt.figure(figsize=(12,8),dpi=100)

axis1 = fig.add_axes([0.1, 0.1, .4, .8])
axis1.plot(x, y)


axis2 = fig.add_axes([0.3, 0.4, .3, .3])
axis2.plot(x, y)
```

Types Of Plots

Types of Plots in Matplotlib

-Plots data ko graphical (visual) form me dikhate hain
-Raw numbers ke bajaye visuals se samajhna easy hota hai

Why use plots:

-Data samajhne ke liye
-Trends aur patterns dekhne ke liye
-Comparison karne ke liye
-Insights nikalne ke liye

Types of plots:

-Line Plot → trend show karta hai
-Scatter Plot → relationship (correlation) dikhata hai
-Bar Plot → categories compare karta hai
-Pie Chart → percentage distribution
-Histogram → data distribution (frequency)
-Box Plot → statistical summary + outliers
-Area Plot → magnitude/volume show karta hai
-Heatmap → patterns via colors

Saving plots

is ka matlab hai jo graph (plot) tum banate ho usko ek file format (PNG, JPG, PDF, SVG) me permanently store karna, taki wo later use, share ya embed kiya ja sake.

Matplotlib me plot pehle memory me Figure object ke form me banta hai. savefig() usi figure ko disk par image/document file me convert karke save karta hai.

Key concepts:

Output formats raster (PNG, JPG) ya vector (PDF, SVG) ho sakte hain
Quality ko resolution (dpi) control karta hai
Layout aur margins ko adjust karke clean output liya ja sakta hai
Saving usually rendering se pehle kiya jata hai taki final state capture ho

```python
fig = plt.figure()
ax =fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```

Image processing,
