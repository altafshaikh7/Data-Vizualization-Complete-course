Matplotlib Basic

subplot ek plotting concept hai jisme ek hi graphical canvas (figure) ko multiple chhote parts me divide kiya jata hai, aur har part me alag data visualize kiya jata hai.

Matlab:
ek figure (poora canvas)
uske andar multiple axes (individual plots)
har axes = ek subplot


subplot ka main goal:

1️⃣ Comparison

Ek hi screen pe multiple graphs compare karna
👉 jaise:

sales vs profit
temperature vs humidity

2️⃣ Space optimization

Alag-alag windows kholne ki jagah
→ ek hi figure me sab dikhao


3️⃣ Data storytelling

Ek hi visualization me multiple aspects dikhana
→ better understanding

```python

x = np.linspace(0, 5, 11)
(x)

y = x**3
(y)
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("x axis")
plt.ylabel("y axis")
```

Subplots in Matplotlin

subplots ek high-level function hai jo ek hi step me:

1. Figure create karta hai
2. Multiple Axes (subplots) create karta hai
3. Unhe structured grid me arrange karta hai

 Matlab:
figure + multiple plots = ek saath ready

 Purpose 
1️⃣ Clean code

Manual switching nahi karna padta (like subplot)

2️⃣ Better control

Har plot ko directly access kar sakte ho:
👉 ax[row, col]

3️⃣ Scalable

Large dashboards, multiple charts easy ho jaate hain

```python
plt.subplot(2, 2, 1)
plt.plot(x,y)
plt.subplot(2, 2, 4)
plt.plot(x,y)
plt.subplot(2, 2, 3)
plt.plot(y,x)
plt.subplot(2, 2, 2)
plt.plot(x**2,x**2)
```

Object Oriented Way

Object Oriented approach me tum directly “objects” (Figure aur Axes) ko control karte ho, instead of global plt functions.

Matlab:

> Tum bolte ho: “is specific plot pe ye draw karo”
> na ki: “jo current plot hai usme draw karo”

Why OO Approach BEST hai?
1️⃣ Full control

Har subplot pe alag settings laga sakte ho

2️⃣ No confusion

subplot() me kaunsa active hai → confusion hota hai
👉 OO me direct access

3️⃣ Scalable

Large projects me clean aur maintainable code


```python
fig = plt.figure()

axis1 = fig.add_axes([0.1, 0.1, .4, .8])
axis1.plot(x, y)

axis2 = fig.add_axes([0.3, 0.4, .3, .3])
axis2.plot(x, y)
```

# 📊 Types of Plots in Matplotlib

## 🧠 What are Plots?

Plots ek graphical (visual) way hai jisme hum data ko charts/graphs ke form me dikhate hain, taki data ko samajhna easy ho jaye.

👉 Instead of numbers (raw data), hum visuals use karte hain.

---

## 🎯 Why we use Plots?

* Data ko easy samajhne ke liye
* Patterns aur trends dekhne ke liye
* Comparison karne ke liye
* Insights nikalne ke liye

---

## 📈 Types of Plots

### 1. Line Plot

* Points ko line se connect karta hai
* Trends show karta hai
* Example: temperature over time

---

### 2. Scatter Plot

* Individual points (dots) dikhata hai
* Relationship aur correlation check karne ke liye
* Example: height vs weight

---

### 3. Bar Plot

* Bars ke form me data
* Categories compare karne ke liye
* Example: sales per product

---

### 4. Pie Chart

* Circle ke parts me data
* Percentage distribution show karta hai
* Example: market share

---

### 5. Histogram

* Data distribution dikhata hai (bins me)
* Frequency check karne ke liye
* Example: exam scores

---

### 6. Box Plot

* Median, quartiles, outliers show karta hai
* Statistical summary ke liye use hota hai

---

### 7. Area Plot

* Filled line chart hota hai
* Quantity ya magnitude show karta hai

---

### 8. Heatmap

* Colors ke through data represent karta hai
* Patterns aur correlations dikhata hai

---

## 🧠 Conclusion

Plots ka main purpose hai data ko visually represent karna taki analysis easy ho jaye.

👉 Right plot choose karna bahut important hota hai.
